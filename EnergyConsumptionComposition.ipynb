```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
data = {
    "Energy_Source": ["Coal", "Natural Gas", "Oil", "Nuclear", "Hydro", "Wind", "Solar", "Biomass"],
    "Percentage": [25, 20, 18, 10, 12, 7, 6, 2]
}
data
```

```python
df = pd.DataFrame(data)
df
```

```python
renewables = {'Hydro', 'Wind', 'Solar', 'Biomass'}
df['Type'] = df['Energy_Source'].apply(lambda x: 'Renewable' if x in renewables else 'Non-renewable')
```

```python
cmap = {
    "Renewable": ["red", "#66C2A5", "#4DB6AC", "#CD8A32"],
    "Non-renewable": ["#922E2E", "#5CCD8B", "#EAFF49", "#64F740"]
}
```

```python
colors = []
for ind, row in df.iterrows():
    palette = cmap[row['Type']]
    group_ind = list(df[df['Type'] == row['Type']].index).index(ind)
    colors.append(palette[group_ind % len(palette)])
```

```python
fig_1, ax_1 = plt.subplots(figsize=(8,8))
wedges, texts, autotexts = ax_1.pie(
    df['Percentage'],
    labels=df['Energy_Source'],
    autopct='%1.1f%%',
    startangle=140,
    colors=colors,
    wedgeprops=dict(width=0.35, edgecolor='white')
)

plt.setp(autotexts, size=10, weight='bold')
plt.setp(texts, size=10)

ax_1.text(0, 0.05, 'Energy\nConsumption', ha='center', va='center', fontsize=14, weight='bold')
ax_1.text(0, -0.12, "Donut Chart", ha='center', va='center', fontsize=10, color='gray')

# Conclusion

renew_sum = df.loc[df["Type"] == "Renewable", "Percentage"].sum()
nonrenew_sum = df.loc[df["Type"] == "Non-renewable", "Percentage"].sum()
summary_text = f"Renewable: {renew_sum}%  |  Non-renewable: {nonrenew_sum}%"
plt.title("Energy Consumption Composition by Source", fontsize=16, pad=20)
```

```python
print("Interpretation:")
print(f"Renewable sources account for {renew_sum:.0f}% of total energy use.")
print(f"Non-renewables make up {nonrenew_sum:.0f}%, indicating heavy fossil fuel dependence.")
print("Increasing renewables would improve sustainability and reduce emissions.")
```
