```python
delay_cols = ['CarrierDelay', 'WeatherDelay', 'NASDelay', 'SecurityDelay', 'LateAircraftDelay']
df[delay_cols] = df[delay_cols].fillna(0)
```

```python
df[delay_cols].isna().sum()
```

```python
# Compute total delay per cause
total_delays = df[delay_cols].sum().sort_values(ascending=False).sum()
print(total_delays)

'''
Delay Type (Total Minutes)

LateAircraftDelay   35,370 <--- When an aircraft arrives late from a previous flight — the biggest reason!
CarrierDelay        15,905 <--- Airline-related (crew, maintenance, scheduling)
NASDelay            4,486  <--- Air traffic control or system delays
WeatherDelay        515    <--- Weather-related delays
SecurityDelay       121    <--- Very small — security checks or alerts
'''
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.bar(total_delays.index, total_delays.values)
plt.title('Total Delay Minutes by Cause')
plt.ylabel('Total Delay (minutes)')
plt.xlabel('Delay Cause')
plt.xticks(rotation=30)
plt.grid(axis='y', alpha=0.7)

'''
Insight from the plot
Late aircraft arrivals and carrier operational issues are the dominant causes of flight delays, together accounting for nearly 90% of total delay time.
'''
```
